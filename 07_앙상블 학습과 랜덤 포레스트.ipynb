## 7.1 투표 기반 분류기

```python
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split

X, y = make_moons(n_samples=500, noise=0.3, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

voting_clf = VotingClassifier(
    estimators=[
        ('lr', LogisticRegression(random_state=42)),
        ('rf', RandomForestClassifier(random_state=42)),
        ('svc', SVC(random_state=42))
    ]
)
voting_clf.fit(X_train, y_train)
```

```python
for name, clf in voting_clf.named_estimators_.items():
    print(name, '=', clf.score(X_test, y_test))
```

```python
voting_clf.predict(X_test[:1])
```

```python
[clf.predict(X_test[:1]) for clf in voting_clf.estimators_]
```

```python
voting_clf.score(X_test, y_test)
```

```python
voting_clf.voting = 'soft'
voting_clf.named_estimators['svc'].probability = True
voting_clf.fit(X_train, y_train)
voting_clf.score(X_test, y_test)
```

## 7.2 배깅과 페이스팅

### 7.2.1 사이킷런의 배깅과 페이스팅

```python
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier

bag_clf = BaggingClassifier(DecisionTreeClassifier(), n_estimators=500, max_samples=100, n_jobs=-1, random_state=42)
bag_clf.fit(X_train, y_train)
```

```python
from matplotlib.colors import ListedColormap

def plot_decision_boundary(clf, X, y, axes, xlabel=r'$x_1$', ylabel=r'$x_2$', alpha=0.3, labels=None):
    custom_cmap = ListedColormap(['#fafab0', '#9898ff', '#a0faa0'])
    colors = ['y', 'b', 'g']
    markers = ['o', '^', 's']
    for i, y_new in enumerate(set(y)):
        plt.scatter(X[y==y_new][:, 0], X[y==y_new][:, 1], color=colors[i], marker=markers[i],
                    s=20, label=labels[i] if labels is not None else None)
    
    x1, x2 = np.meshgrid(np.linspace(axes[0], axes[1], 100),
                         np.linspace(axes[2], axes[3], 100))
    X_new = np.c_[x1.ravel(), x2.ravel()]
    y_pred = clf.predict(X_new).reshape(x1.shape)
    plt.contourf(x1, x2, y_pred, cmap=custom_cmap, alpha=alpha)
            
    plt.axis(axes)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    if labels is not None:
        plt.legend()
```

```python
tree_clf = DecisionTreeClassifier(random_state=42)
tree_clf.fit(X_train, y_train)

axes = [-1.5, 3, -1, 1.5]
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.title('Decision tree')
plot_decision_boundary(tree_clf, X_test, y_test, axes=axes)
plt.subplot(1, 2, 2)
plt.title('Decision tree with bagging')
plot_decision_boundary(bag_clf, X_test, y_test, axes=axes)
plt.show()
```

## 7.5 부스팅

### 7.5.1 AdaBoost

```python
m = len(X_train)
axes = [-1.5, 3, -1, 1.5]

plt.figure(figsize=(10, 4))
for subplot, lr in ((1, 1), (2, 0.5)):
    sample_weights = np.ones(m) / m
    
    for i in range(5):
        svm_clf = SVC(C=0.2, gamma=0.6, random_state=42)
        svm_clf.fit(X_train, y_train, sample_weight=sample_weights * m)
        y_pred = svm_clf.predict(X_train)

        r = sample_weights[y_pred != y_train].sum() / sample_weights.sum()
        alpha = lr * np.log((1 - r) / r)
        sample_weights[y_pred != y_train] *= np.exp(alpha)
        sample_weights /= sample_weights.sum()

        plt.subplot(1, 2, subplot)
        plt.title(f'learning rate = {lr}')
        plot_decision_boundary(svm_clf, X_train, y_train, axes, alpha=0.2)
        
plt.show()
```

### 7.5.2 그레이디언트 부스팅

```python
from sklearn.tree import DecisionTreeRegressor

np.random.seed(42)
X = np.random.rand(100, 1) - 0.5
y = 3 * X[:, 0] ** 2 + 0.05 * np.random.randn(100)
```

```python
tree_reg1 = DecisionTreeRegressor(max_depth=2, random_state=42)
tree_reg1.fit(X, y)
```

```python
y2 = y - tree_reg1.predict(X)
tree_reg2 = DecisionTreeRegressor(max_depth=2, random_state=43)
tree_reg2.fit(X, y2)
```

```python
y3 = y2 - tree_reg2.predict(X)
tree_reg3 = DecisionTreeRegressor(max_depth=2, random_state=44)
tree_reg3.fit(X, y3)
```

```python
X_new = np.array([[-0.4], [0.], [0.5]])
sum(tree.predict(X_new) for tree in (tree_reg1, tree_reg2, tree_reg3))
```

```python
trees = [tree_reg1, tree_reg2, tree_reg3]
X_new = np.linspace(-0.5, 0.5, 200).reshape(-1, 1)

plt.figure(figsize=(10, 10))

for i, ylabel in enumerate(['$y$', '$y - h_1(x_1)$', '$y - h_1(x_1) - h_2(x_1)$']):
    plt.subplot(3, 2, 2 * i + 1)
    if i == 0:
        plt.scatter(X, y, s=10, label='Training set')
        plt.xlabel('$x_1$')
    else:
        y_new = y - sum(tree.predict(X) for tree in trees[:i])
        plt.scatter(X, y_new, s=20, marker='+', c='k', label='Residuals')
        plt.ylim(-0.5, 0.5)
    plt.plot(X_new, trees[i].predict(X_new), c='green', label=f'$h_{i+1}(x_1)$')
    plt.xlim(-0.5, 0.5)
    plt.ylabel('$y$')
    plt.legend(loc='upper center')

for i, label in enumerate(['$h(x_1) = h_1(x_1)$', '$h(x_1) = h_1(x_1) + h_2(x_1)$',
                           '$h(x_1) = h_1(x_1) + h_2(x_1) + h_3(x_1)$']):
    plt.subplot(3, 2, 2 * i + 2)
    plt.scatter(X, y, s=10, label='Training set')
    y_pred = sum(tree.predict(X_new) for tree in trees[:i+1])
    plt.plot(X_new, y_pred, c='red', label=label)

    if i == 0:
        plt.title('Ensemble predictions')
    if i == 2:
        plt.xlabel('$x_1$')
    plt.ylabel('$y$')
    plt.xlim(-0.5, 0.5)
    plt.legend(loc='upper center')
    
plt.show()
```

```python
from sklearn.ensemble import GradientBoostingRegressor

plt.figure(figsize=(10, 4))
for i, lr, n_estimators in ((1, 1.0, 3), (2, 0.1, 200)):
    gbrt = GradientBoostingRegressor(learning_rate=lr, n_estimators=n_estimators, random_state=42)
    gbrt.fit(X, y)
    
    plt.subplot(1, 2, i)
    plt.title(f'learning_rate={lr}, n_estimators={n_estimators}')
    plt.scatter(X, y, s=10)
    plt.plot(X_new, gbrt.predict(X_new), c='r')
    plt.xlim(-0.5, 0.5)
    plt.xlabel('$x_1$')
    plt.ylabel('$y$')

plt.show()
```

```python
gbrt_best = GradientBoostingRegressor(max_depth=2, learning_rate=0.05, n_estimators=500,
                                      n_iter_no_change=10, random_state=42)
gbrt_best.fit(X, y)

gbrt_best.n_estimators_
```

```python
plt.scatter(X, y)
plt.plot(X_new, gbrt.predict(X_new), c='r')
plt.xlim(-0.5, 0.5)
plt.xlabel('$x_1$')
plt.ylabel('$y$')
plt.show()
```

### 7.5.3 히스토그램 기반 그레이디언트 부스팅

```python
from pathlib import Path
import pandas as pd
import tarfile
import urllib.request

def load_housing_data():
    tarball_path = Path("datasets/housing.tgz")
    if not tarball_path.is_file():
        Path("datasets").mkdir(parents=True, exist_ok=True)
        url = "https://github.com/ageron/data/raw/main/housing.tgz"
        urllib.request.urlretrieve(url, tarball_path)
        with tarfile.open(tarball_path) as housing_tarball:
            housing_tarball.extractall(path="datasets")
    return pd.read_csv(Path("datasets/housing/housing.csv"))

housing = load_housing_data()

train_set, test_set = train_test_split(housing, test_size=0.2, random_state=42)
housing_labels = train_set["median_house_value"]
housing = train_set.drop("median_house_value", axis=1)
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.compose import make_column_transformer
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.preprocessing import OrdinalEncoder

hgb_reg = make_pipeline(
    make_column_transformer((OrdinalEncoder(), ['ocean_proximity']), remainder='passthrough'),
    HistGradientBoostingRegressor(categorical_features=[0], random_state=42)  # 결측값, 스케일 조정, 원핫인코딩 불필요
)
hgb_reg.fit(housing, housing_labels)
```

```python
from sklearn.model_selection import cross_val_score

hgb_rmse = -cross_val_score(hgb_reg, housing, housing_labels, scoring='neg_root_mean_squared_error', cv=10)
pd.Series(hgb_rmse).describe()
```

```python
from sklearn.datasets import fetch_openml

X_mnist, y_mnist = fetch_openml('mnist_784', return_X_y=True, as_frame=False)
```

```python
X_train, y_train = X_mnist[:50_000], y_mnist[:50_000]
X_valid, y_valid = X_mnist[50_000:60_000], y_mnist[50_000:60_000]
X_test, y_test = X_mnist[60_000:], y_mnist[60_000:]
```

```python
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.neural_network import MLPClassifier

rnd_forest_clf = RandomForestClassifier(random_state=42)
extra_trees_clf = ExtraTreesClassifier(random_state=42)
mlp_clf = MLPClassifier(random_state=42)
```

```python
rnd_forest_clf.fit(X_train, y_train)
extra_trees_clf.fit(X_train, y_train)
mlp_clf.fit(X_train, y_train)

estimators = [rnd_forest_clf, extra_trees_clf, mlp_clf]
for estimator in estimators:
    print(estimator, '\t:', estimator.score(X_valid, y_valid))
```

```python
named_estimators = [
    ("random_forest_clf", rnd_forest_clf),
    ("extra_trees_clf", extra_trees_clf),
    ("mlp_clf", mlp_clf),
]

voting_clf = VotingClassifier(named_estimators)
voting_clf.fit(X_train, y_train)
```

```python
voting_clf.score(X_valid, y_valid)
```

```python
voting_clf.voting = 'soft'
voting_clf.score(X_valid, y_valid)
```
